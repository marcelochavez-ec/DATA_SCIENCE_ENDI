```python
def suma_numeros(*args, **kwargs):
    """
    Suma los números dados a través de *args y **kwargs. 
    Los argumentos pueden ser pasados como números en *args 
    o con claves en **kwargs (por ejemplo, 'num1', 'num2', etc.).
    
    :param args: números a sumar de manera posicional.
    :param kwargs: números a sumar a través de claves.
    :return: la suma total de los números.
    """
    
    # Validar que los valores sean numéricos
    total = 0
    for num in args:
        if not isinstance(num, (int, float)):
            raise ValueError(f"El valor {num} en *args no es un número válido.")
        total += num
    
    for key, num in kwargs.items():
        if not isinstance(num, (int, float)):
            raise ValueError(f"El valor para '{key}' en **kwargs no es un número válido.")
        total += num
    
    # Impresión sofisticada
    print(f"Los números a sumar son: {', '.join(map(str, args)) + ', ' if args else ''}" + \
          ', '.join([f"{key}: {num}" for key, num in kwargs.items()]))
    print(f"La suma total es: {total}")
    
    return total
```

```python
# Usando *args
suma_numeros(1, 2, 3, 4)

# Usando **kwargs
suma_numeros(num1=1, num2=2, num3=3)

# Combinando *args y **kwargs
suma_numeros(1, 2, 3, num4=4, num5=5)
```
