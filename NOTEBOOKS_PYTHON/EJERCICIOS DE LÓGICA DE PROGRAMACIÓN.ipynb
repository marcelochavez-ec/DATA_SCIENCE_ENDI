1. Inicio
2. Sea I una variable cuyo valor inicial es 1 se asina a I
3. Calcular el producto de I por I y llamar J al resultado
4. Calcular el producto de I por J y llamar K al resultado
5. Escribir los valores de I,J y K, en la misma línea la hoja de respuestas
6. Aumentar en 1 el valor de la variable I y asignar a I
7. Si I-99 es igual o menor a cero, volver al paso 3, sino ir al paso 8
8. Fin del algoritmo

```python
# 1. Inicio
i = 1  # 2. Inicializar la variable i en 1

print(f"{'Número':<10}{'Cuadrado':<10}{'Cubo':<10}")  # Encabezado de la tabla
print("=" * 30)

while i - 99 <= 0:  # 7. Si i - 99 es menor o igual a 0, continuar el ciclo
    j = i * i  # 3. Calcular el cuadrado y asignarlo a j
    k = i * j  # 4. Calcular el cubo y asignarlo a k
    print(f"{i:<10}{j:<10}{k:<10}")  # 5. Escribir los valores de i, j y k
    i = i + 1  # 6. Aumentar i en 1

# 8. Fin del algoritmo
```

```python
# 1. Inicio
i = 1  # 2. Inicializar la variable i en 1
valores = []  # Lista para almacenar los resultados

while i - 99 <= 0:  # 7. Si i - 99 es menor o igual a 0, continuar el ciclo
    j = i * i  # 3. Calcular el cuadrado y asignarlo a j
    k = i * j  # 4. Calcular el cubo y asignarlo a k
    valores.append((i, j, k))  # Guardar valores en la lista
    i = i + 1  # 6. Aumentar i en 1

# 5 primeros valores
print(f"{'Número':<10}{'Cuadrado':<10}{'Cubo':<10}")  # Encabezado de la tabla
print("=" * 30)
for num, cuadrado, cubo in valores[:5]:  # Primeros 5 elementos
    print(f"{num:<10}{cuadrado:<10}{cubo:<10}")

print("...")  # Indicación de que se omitieron valores intermedios

# 5 últimos valores
for num, cuadrado, cubo in valores[-5:]:  # Últimos 5 elementos
    print(f"{num:<10}{cuadrado:<10}{cubo:<10}")
```
