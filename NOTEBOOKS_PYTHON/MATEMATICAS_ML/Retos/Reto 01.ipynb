# Ejercicio 1
Resolver los siguientes sistemas de ecuaciones

a)\
$\begin{array}{rll}x_1&+&x_2&+&x_3&=&3\\x_1&-&x_2&+&2x_3&=&2\\2x_1&&&+&3x_3&=&1\end{array}$

```python
# Construir el sistema de ecuaciones de forma vectorial
  
A = np.vstack([[1,1,1],[1,-1,2],[2,0,3]])
b = np.vstack([3,2,1])
print(A)
print(b)
```

```python
try:
  npl.solve(A,b)
except:
  print(np.hstack([A,b]))
  print("No tiene solución")
```

b)\
$\begin{array}{rll}x_1&+&x_2&+&x_3&=&3\\x_1&-&x_2&+&2x_3&=&2\\&&x_2&+&x_3&=&2\end{array}$

```python
A = np.vstack([[1,1,1],[1,-1,2],[2,0,3]])
b = np.vstack([3,2,5])
print(np.hstack([A,b]))
print("La solucion es")
npl.lstsq(A,b, rcond=None)[3]
```

c)\
$\begin{array}{rll}x_1&+&x_2&+&x_3&=&3\\x_1&-&x_2&+&2x_3&=&2\\2x_1&&&+&3x_3&=&5\end{array}$

```python
A = np.vstack([[1,1,1],[1,-1,2],[2,0,3]])
b = np.vstack([3,2,5])
print(np.hstack([A,b]))
print("La solucion es")
npl.lstsq(A,b,rcond= None)[3] # Calcula la solucion por minimos cuadrados
```

# Ejercicio 2
**Interpretacion Geometrica de Sistemas Lineales de Ecuaciones**

En un sistema lineal de ecuaciones con dos variables $x_1$ y $x_2$, cada ecuacion lineal define una linea en el plano $x_1x_2$. Debido a que la solucion al sistema de ecuaciones debe satisfacer todas las ecuaciones de manera simultanea, entonces el conjunto solucion es la interseccion de esas lineas. Por ejemplo, considere el siguiente sistema de ecuaciones
$$\begin{array}{rll} 4x_1+4x_2&=5\\2x_1-4x_2&=1\end{array}$$
cuyo espacio de solucion es el punto $(x_1,x_2)=(1,\frac{1}{4})$.

```python
import matplotlib.pyplot as plt

vals = np.linspace(-10,10,10000) # Valores de prueba

eq0 = lambda x1: (5-4*x1)/4 # 4x1+4x2 =5
y0 = eq0(vals)

eq1 = lambda x1: (2*x1-1)/4 # 2x1-4x2 =1
y1 = eq1(vals)

plt.plot(vals,y0, c="blue")
plt.plot(vals,y1, c="red")
idx = np.argwhere(np.diff(np.sign(y0-y1))).flatten() #Encontrar la interseccion entre las dos graficas
plt.plot(vals[idx],y0[idx],'ko')
plt.legend(["$4x_1+4x_2=5$","$2x_1-4x_2=1$",
            str(np.round(vals[idx],1))+str(np.round(y0[idx],2))])
plt.show()
```
