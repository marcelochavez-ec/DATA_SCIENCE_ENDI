**Resolviendo Sistemas de Ecuaciones Lineales**

Considere el siguiente sistema de ecuaciones
$$\begin{bmatrix}
1&0&8&-4\\
0&1&2&12
\end{bmatrix}\begin{bmatrix}x_1\\x_2\\x_3\\x_4\end{bmatrix}=\begin{bmatrix}48\\8\end{bmatrix}$$


```python
# Codificar las matrices como Ax=b
A = np.matrix([[1,0,8,-4],[0,1,2,12]])
b = np.vstack([42,8])
A,b
```

Una forma para resolver el sistema de ecuaciones es aprovechando que se tiene una matriz identidad en las primeras dos variables. Entonces el vector $\textbf{b}$ puede ser escrito como
$$\textbf{b}=\begin{bmatrix}42\\8\end{bmatrix}=42\begin{bmatrix}1\\0\end{bmatrix}+8\begin{bmatrix}0\\1\end{bmatrix}$$
Entonces una posible solucion es $\textbf{x}=[42,8,0,0]^\top$. Esta solucion se le conoce como solucion particular.

```python
# Verifiquemos que x=[42,8,0,0]^T es una solucion al sistema de ecuaciones
sol1 = np.dot(A,np.vstack([42,8,0,0]))
print(sol1)
```

Para obtener todas las soluciones, se tiene que ser creativo en generar un **0** en una forma no trivial usando las columnas de la matriz. Para eso se expresara la tercer columna en terminos de las primeras dos columnas, es decir,
$$\begin{bmatrix}8\\2\end{bmatrix}=8\begin{bmatrix}1\\0\end{bmatrix}+2\begin{bmatrix}0\\1\end{bmatrix}$$
tal que $\textbf{0}=8\textbf{c}_1+2\textbf{c}_2-1\textbf{c}_3+0\textbf{c}_4$ y $(x_1,x_2,x_3,x_4)=(8,2,-1,-0)$. En realidad, cualquier escalamiento a esta solucion por un escalar $\lambda_1\in\mathbb{R}$ produce el vector $\textbf{0}$, i.e., 
$$\begin{bmatrix}1&0&8&-4\\0&1&2&12\end{bmatrix}\left(\lambda_1\begin{bmatrix}8\\2\\-1\\0\end{bmatrix}\right)=\lambda_1(8\textbf{c}_1+2\textbf{c}_2-\textbf{c}_3)=\textbf{0}$$

```python
# Tomemos un escalar lambda1 aleatorio
lambda1 = 4815162342
zero1 = np.dot(A,lambda1*np.vstack([A[:,2],-1,0]))
print(zero1)
```

Siguiendo un procedimiento similar para la cuarta columna se obtiene que
$$\begin{bmatrix}1&0&8&-4\\0&1&2&12\end{bmatrix}\left(\lambda_2\begin{bmatrix}-4\\12\\0\\-1\end{bmatrix}\right)=\lambda_2(-4\textbf{c}_1+12\textbf{c}_2-\textbf{c}_4)=\textbf{0}$$
para cualquier $\lambda_2\in\mathbb{R}$. Entonces la solucion general al sistema de ecuaciones es
$$\left\{\textbf{x}\in\mathbb{R}^4:=\textbf{x}=\begin{bmatrix}42\\8\\0\\0\end{bmatrix}+\lambda_1 \begin{bmatrix}8\\2\\-1\\0\end{bmatrix}+\lambda_2 \begin{bmatrix}-4\\12\\0\\-1\end{bmatrix}, \lambda_1\lambda_2\in\mathbb{R}\right\}.$$

```python
# La cuarta columna es expresada en terminos de las primeras dos
# Tomemos un escalar lambda2 arbitrario
lambda2 = -1678906543
zero2= np.dot(A,lambda2*np.vstack([A[:,3],0,-1]))
print(zero2)
```

Debido a que los dos nuevos vectores encontrados son **0** entonces al sumarlo a la solucion particular arroja el mismo resultado, dando como resultado la solucion general.

```python
# Por lo tanto la solucion general es
xsol = np.vstack([42,8,0,0])+lambda1*np.vstack([A[:,2],-1,0])+lambda2*np.vstack([A[:,3],0,-1])
sol = np.dot(A,xsol)
print(sol)
```
