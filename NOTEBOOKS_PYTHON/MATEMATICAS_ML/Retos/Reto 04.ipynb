Efectivamente, la matriz se encuentra en su forma escalonada reducida. Para obtener la solucion al sistema de ecuaciones homogeneo se puede seguir dos caminos. 
1.   Escribir las columnas pivote en terminos de las columnas que no son pivote.
2.   Buscar obtener un **0** no trivial.

Para el primer caso es sencillo observar que
$$\begin{array}{rll}
x_1+3x_2+3x_5&=0\\
x_3+9x_5&=0\\
x_4-4x_5&=0
\end{array}$$

De donde inmediatamente se obtiene que la solucion general es
$$\left\{\textbf{x}\in\mathbb{R}^5: \textbf{x}=\lambda_1\begin{bmatrix}-3\\1\\0\\0\\0\end{bmatrix}+\lambda_2\begin{bmatrix}-3\\0\\-9\\4\\1\end{bmatrix}\ \lambda_1,\lambda_2\in\mathbb{R}\right\}$$
