```python
import numpy as np
import matplotlib.pyplot as plt
```

Observe si los siguientes conjuntos son subespacios en $\mathbb{R}^2$ de $V$.

```python
# Caso A un cuadrilatero en R2
origen = np.vstack([0,0])  # Origen de coordenadas
v = 1.5                    # Factor de escala
scale = v*10               # Escalamiento
vec = np.array([v,0])      # Vector 0
vec1 = np.array([1,1])     # Vector 1
vec2 = np.array([0,1])     # Vector 2
plt.axis([-scale/2,scale/2,-scale/2,scale/2])
# Conjunto a analizar
plt.fill_between(np.linspace(-v,v,100),np.linspace(v,v,100),np.linspace(-v,-v,100), color ="pink")
# Graficar los 3 vectores propuestos
plt.quiver(*origen, *vec, scale = scale)
plt.quiver(*origen, *vec1, scale = scale, color = "gold")
plt.quiver(*origen, *vec2, scale = scale, color = "green")
plt.quiver(*origen, *np.array(vec1+vec2), scale = scale,color = "red") 
plt.show()
```

```python
# Caso B, ecuacion de una linea en R2
# Se grafica la linea y = x+2
plt.plot(np.linspace(-6,6,100),np.linspace(-6,4,100)+2, color ="red")

# Se grafica el plano Cartesiano
plt.plot(np.linspace(-6,6,100),np.linspace(0,0,100),color="k")
plt.plot(np.linspace(0,0,100),np.linspace(-6,6,100),color="k")
plt.scatter(0,0,color="k")  # Origen de coordenadas
plt.show()
```

```python
# Caso C, el conjunto generado entre dos rectas

x = np.linspace(-6,6,100)        # Valor de la ordenada
y1 = np.linspace(-6*2,6*2,100)   # Recta 1
y2 = np.linspace(-6*5,6*5,100)   # Recta 2

# Graficar las rectas, el orgen de coordenadas y las coordenadas
plt.plot(x,y1, color = "k")
plt.plot(x,y2, color = "k")
plt.plot(x,np.linspace(0,0,100),color="k") 
plt.plot(np.linspace(0,0,100),np.linspace(-30,30,100),color="k") 
plt.scatter(0,0,color="k") 

# Colorear el conjunto generado en la interseccion de las rectas
plt.fill_between(x,y1,y2, color = 'pink') # Conjunto a analizar
v1 = np.array([3,1.4])    # Usar vectores aleatorios
v2 = np.array([-2,-1.3]) 
v3 = 4*(v1+v2)

plt.quiver(*origen,*v1,scale = scale, color = "red") # v1
plt.quiver(*origen,*v2,scale = scale, color = "blue") # v2
plt.quiver(*origen,*v3,scale = scale, color = "gold") # v3
plt.legend(["$v_1$:"+str(v1),"$v_2$:"+str(v2),"$v_3$:"+str(v3)])
plt.show()
```

```python
# Cado D, un punto en R2
plt.scatter(0,0)   # Punto localizado en el origen
plt.plot(x,np.linspace(0,0,100),color="k") 
plt.plot(np.linspace(0,0,100),np.linspace(-30,30,100),color="k") 
plt.show()
```
