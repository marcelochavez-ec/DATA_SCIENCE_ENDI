# Ejercicio 1
Una base estandar canonica en $\mathbb{R}^3$ es
$$\mathcal{B}=\left\{\begin{bmatrix}1\\0\\0\end{bmatrix},\begin{bmatrix}0\\1\\0\end{bmatrix},\begin{bmatrix}0\\0\\1\end{bmatrix}\right\}$$

Otras bases diferentes en $\mathbb{R}^3$ son
$$\mathcal{B}_1=\left\{\begin{bmatrix}1\\1\\1\end{bmatrix},\begin{bmatrix}0\\1\\1\end{bmatrix},\begin{bmatrix}0\\0\\1\end{bmatrix}\right\}$$
$$\mathcal{B}_2=\left\{\begin{bmatrix}0.5\\1.8\\-2.2\end{bmatrix},\begin{bmatrix}0.8\\0.3\\-1.3\end{bmatrix},\begin{bmatrix}0.4\\0.3\\3.5\end{bmatrix}\right\}$$

```python
# Graficar las bases
origen = np.array([0,0,0])
B = np.vstack([[1,0,0],[0,1,0],[0,0,1]]) 
B1 = np.vstack([[1,1,1],[0,1,1],[0,0,1]]) 
B2 = np.vstack([[.5,1.8,-2.2],[.8,.3,-1.3],[.4,.3,3.5]]) 
bases = [B,B1,B2]
print(B)
print(B1)
print(B2)
```

```python
from mpl_toolkits.mplot3d import Axes3D
plt.rcParams[ "figure.figsize" ] = (8,8) 

fig = plt.figure()
ax = fig.gca(projection = '3d')
scale3D = 1
ax.set_xlim3d(-scale3D,scale3D)
ax.set_ylim3d(scale3D,-scale3D)
ax.set_zlim3d(-scale3D,scale3D)

c = ["r","g","b"]
lw = [3,1.5,.25]
for ind,base in enumerate(bases):
    for j,v in enumerate(base):
        ax.quiver(*origen,*base[:,j],length=scale3D/2, color=c[ind], linewidths=lw[ind])

     
plt.show()
```

En pocas palabras, una base sirve para definir un vector bajo una perspectiva diferente.
