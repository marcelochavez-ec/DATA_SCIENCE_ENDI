# Ejercicio 1
Es bien sabido que el sistema de coordenadas Cartesiano $\{\textbf{e}_1=[1,0]^\top,\textbf{e}_2=[0,1]^\top\}$ es una base canónica en $\mathbb{R}^2$. En este sistema de coordenadas, un vector $\textbf{x}\in\mathbb{R}^2$ tiene una representación que dice como combinar linealmente $\textbf{e}_1$ y $\textbf{e}_2$ para obtener $\textbf{x}$. Sin embargo, toda base en $\mathbb{R}^2$ define un sistema de coordenadas válido y entonces el mismo vector $\textbf{x}$ tiene una representación distinta en este nuevo sistema de coordenadas con base $(\textbf{b}_1,\textbf{b}_2)$.

```python
plt.rcParams[ "figure.figsize" ] = (5,5)
origen = np.array([0,0])
e_base = np.vstack([[1,0],[0,1]])           # Base Canónica en el sistema de coordenadas Cartesiano
b_base = np.vstack([[1.5,.75],[.25,1.5]])   # Base aleatoria en R2

# Se crea un vector v=[2,2] que será modificado acorde a las bases
v1 = lambda b: np.dot(b,np.array([2,2]))
scale = 10    # Para visualización

plt.axis([-scale/2,scale/2,-scale/2,scale/2])     # Espacio 2D
plt.quiver(*origen,*v1(e_base),scale=scale,color="b")  # Vector v1 visto desde la base e_base
plt.quiver(*origen,*v1(b_base),scale=scale,color="r")  # Vector v1 visto desde la base b_base
plt.quiver(*origen,*e_base[:,0],scale=scale,color="k")   # Gráfica e_base
plt.quiver(*origen,*e_base[:,1],scale=scale,color="k")   # Gráfica e_base
plt.quiver(*origen,*b_base[:,0],scale=scale,color="g")   # Gráfica b_base
plt.quiver(*origen,*b_base[:,1],scale=scale,color="g")   # Gráfica b_base
plt.grid()
plt.show()
```
