Un subespacio afín de una dimensión se le denomina línea y puede ser escrito como $\textbf{y}=\textbf{x}_0+\lambda\textbf{x}_1$, donde $\lambda\in\mathbb{R}$ donde $U=\text{span}[x_1]\subseteq\mathbb{R}^n$ es un subespacio de una sola dimensión de $\mathbb{R}^n$.

```python
# Graficar un espacio afín
scale = 10
origen = np.array([0,0])
plt.rcParams[ "figure.figsize" ] = (10,10)
plt.axis([-scale/2,scale/2,-scale/2,scale/2])


# Data setup.
lambda1 = 1
x0 = np.___([-1,1])       # Punto de soporte
x1 = np.___([3,.5])       # Punto que define la dirección
val = x0+___*x1          # Línea
x1s = np.vstack([k*x1 for k in np.linspace(-4,4,250)]) # Crear una línea en dirección de x1

# Graficando espacio afín
val2 = np.vstack([x0+l*x1 for l in np.linspace(-4,4,250)]) 
plt.plot(x1s[:,0],x1s[:,1],color="b")
plt.plot(val2[:,0],val2[:,1],color="r")

# Graficar vectores
plt.___(*origin, *x1, scale=scale, color="k")
plt.___(*origin, *x0, scale=scale, color="blueviolet")
plt.___(*origin, *val, scale=scale, color="orange") 

plt.legend(["$x_1\lambda$","L = x_0 + \lambda x_1$","$x_1$","$x_0$","$y$"])
plt.show()
```
